# Brand And Category Analysis

## Top Brand

```python
print(df['Brand'].value_counts().nlargest(10))
```

```python
brand_counts = df.Brand.value_counts().reset_index().head(10)
brand_counts.columns = ['Brand','Product_Count']

plt.figure(figsize=(15,8))
plt.bar(brand_counts["Brand"], brand_counts["Product_Count"], color=sb.color_palette("ch:start=.2,rot=-.3", len(brand_counts)),edgecolor='black')
plt.title("Top 10 Brands by Number of Products")
plt.xlabel("Brand")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Which Brand have more Product 

### Performax has the most products, with approximately 3,000. Teamspirit, DNMX, and Netplay follow, while Lee Cooper has the fewest products among the top 10 brands.


# Category Analysis

```python
category_counts = df["Category"].value_counts().reset_index()
category_counts.columns = ["Category", "Product_Count"]

plt.figure(figsize=(14,7))
plt.pie(category_counts["Product_Count"], labels=category_counts["Category"], autopct='%1.1f%%', startangle=180,
           pctdistance=0.85,labeldistance=1.1,wedgeprops={'linewidth':1, 'edgecolor':'white'},
           colors=sb.color_palette("ch:start=.2,rot=-.3", len(category_counts)))
plt.title("Product Distribution across Category")
plt.show()
```

## Which Category dominates overall 

### Topwear is the largest category, accounting for 22.5% of the products, followed by Footwear (19.1%). This shows that Topwear dominates the overall product distribution in the dataset.

# Category Vs Price Segment

```python
price_category_counts = df.groupby(['Price_Segment','Category']).size().reset_index(name="Count")

plt.figure(figsize=(14,7))
sb.barplot(x=price_category_counts.Price_Segment, y=price_category_counts.Count, 
            hue=price_category_counts.Category, palette="ch:start=.2,rot=-.3",edgecolor='black')

plt.title("Category Domination Across Price Segments")
plt.xlabel("Price Segment")
plt.ylabel("Number of Products")
plt.legend(title="Category", bbox_to_anchor=(1.0, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Which Category dominates in Budget , Mid and Premium Luxury

### The Mid-Range segment has the highest number of products, especially in Topwear and Westernwear. In the Premium segment, Footwear has the highest count. The Low segment is also dominated by Footwear, while the Luxury segment has very few products overall.

# Discount % Vs Product Type

```python
avg_discount = df.groupby("Product_Name")["Discount_Percentage"].mean().reset_index().sort_values("Discount_Percentage", ascending=False).head(15)

plt.figure(figsize=(15,7))
sb.barplot(x=avg_discount.Product_Name, y=avg_discount.Discount_Percentage, hue=avg_discount.Product_Name, palette='coolwarm', edgecolor='black', legend=False)
plt.title("Top 10 Product Types with Highest Average Discount")
plt.xlabel("Product Type")
plt.ylabel("Average Discount (%)")
plt.xticks(rotation=45)
plt.show()
```

### The Brocade Tunic,  Brocade Angarakha, Chanderi Kurta, and Chanderi Top have the highest average discount, at approximately 80% each.

# Price Segment Analysis

## Distribution of products across Budget, Mid, Premium, Luxury

```python
segment_counts = df["Price_Segment"].value_counts().reset_index()
segment_counts.columns = ["Price_Segment", "Product_Count"]

plt.figure(figsize=(8,7))
plt.pie(segment_counts["Product_Count"], labels=segment_counts["Price_Segment"], autopct='%1.1f%%',
        pctdistance=0.85,labeldistance=1.1,wedgeprops={'linewidth':1, 'edgecolor':'white'},
        startangle=140,colors=sb.color_palette("mako", len(segment_counts)))
plt.title("Product Distribution Across Price Segment")
plt.ylabel('Price Segment')
plt.show()
```

```python
category_by_segment = df.groupby(["Price_Segment", "Category"]).size().reset_index(name="Count").sort_values(["Price_Segment","Count"], ascending=[True, False])

plt.figure(figsize=(14,7))
sb.barplot(x=category_by_segment.Count, y=category_by_segment.Category,
            hue=category_by_segment.Price_Segment,dodge=True,palette="mako",edgecolor='white')
plt.title("Top Categories Across Price Segments")
plt.xlabel("Number of Products")
plt.ylabel("Category")
plt.legend(title="Price Segment", bbox_to_anchor=(1.0,1), loc="upper left")
plt.tight_layout()
plt.show()
```

### The Mid-Range segment (₹500–₹1499) has the highest number of products, with Topwear leading, followed by Westernwear. In the Premium segment, Footwear is the leading category. Overall, most products are concentrated in the Mid-Range segment, while Luxury has the fewest products.
